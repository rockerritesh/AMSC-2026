```python
t_start = 0.0
t_end = 0.5 * 3.145
N = 500000

h = (t_end - t_start) / N
```

```python
y1 = 0.0
y2 = 1.0

t = t_start
```

```python
y_values = [[y1, y2]]
t_values = [t_start]
```

```python
for i in range(N):
    dy1=y2
    dy2=-y1
    y1 += h*dy1
    y2 += h*dy2
    t += h

    y_values.append([y1, y2])
    t_values.append(t)
```

```python
print(t)
print(y1, y2)
print(y1**2 + y2**2)
```

```python
import matplotlib.pyplot as plt
plt.plot(t_values, [y[0] for y in y_values], label='y1')
plt.plot(t_values, [y[1] for y in y_values], label='y2')
plt.xlabel('Time')
plt.ylabel('Values')
```
